```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('DataCoSupplyChainDataset.csv', encoding='latin1')
df.head()#shows first 5 rows of the dataset
```

```python
df.describe()#gives the statistical summary of columns
```

```python
df.info()#shows null values and data types of columns
```

```python
df.isna().sum()#sum of null values in each column
```

```python
#Dropping null columns
df = df.drop(['Order Zipcode', 'Product Description'], axis=1)
```

```python
#Drop unwanted & unnecessary columns
df = df.drop(['Customer Lname','Customer Fname','Customer Email', 'Customer Password'], axis=1)
```

```python
#Drop noise columns
df = df.drop(['Product Image','Product Status'], axis=1)
```

```python
#Drop rows with null values of the customer zipcode
df = df.dropna(subset=['Customer Zipcode'])
```

```python
#check data types of columns
df.dtypes
```

```python
df.isna().sum()#check null values again
```

```python
#check correlation between features to drop duplicate columns with different names to avoid noise and data leakage
abs(df.corr(numeric_only=True)['Category Id']).sort_values(ascending=False)
```

```python
#encoding all categorical values
from sklearn.preprocessing import LabelEncoder
categorical_cols = df.select_dtypes(include=['object']).columns#choose object columns
for col in categorical_cols:
    le = LabelEncoder()#create le as an object of the labelencoder class
    df[col] = le.fit_transform(df[col].astype(str))#encodes categorical values
```

```python
#check data types
df.dtypes
```

```python
#find correlation between features
corr_matrix = df.corr()

plt.figure(figsize=(24, 22))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap of Input Features', fontsize=18)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
corr_matrix
```

```python
#drop columns that have the same values but different name
df = df.drop(['Benefit per order','Order Item Total', 'Product Category Id','Order Item Cardprod Id', 'Customer Id','Product Price','Order Id'], axis=1)
```

```python
abs(df.corr(numeric_only=True)['Order Profit Per Order']).sort_values(ascending=False)
```

```python
abs(df.corr(numeric_only=True)['Late_delivery_risk']).sort_values(ascending=False)
```

```python
#boxplot for order profit per order to check if we have outliers
plt.figure(figsize=(12, 6))
sns.boxplot(x=df['Order Profit Per Order'], color="skyblue")
plt.title("Boxplot of Order Profit Per Order", fontsize=16)
plt.xlabel("Order Profit Per Order", fontsize=14)
plt.show()
```

```python
df.columns
```

```python
import pandas as pd

# Calculate Q1, Q3, and IQR for the profit column
Q1 = df['Order Profit Per Order'].quantile(0.25)
Q3 = df['Order Profit Per Order'].quantile(0.75)
IQR = Q3 - Q1

# Define outlier thresholds
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Extract all outliers
outliers = df[(df['Order Profit Per Order'] < lower_limit) | (df['Order Profit Per Order'] > upper_limit)]

upper_outliers = outliers[outliers['Order Profit Per Order'] > upper_limit]
lower_outliers = outliers[outliers['Order Profit Per Order'] < lower_limit]

#found a huge amount of outliers so decided to keep them
print(f"Upper Outliers: {len(upper_outliers)}")
print(f"Lower Outliers: {len(lower_outliers)}")
```
